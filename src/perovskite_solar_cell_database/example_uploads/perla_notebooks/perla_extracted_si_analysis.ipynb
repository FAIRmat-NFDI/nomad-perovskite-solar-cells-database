```python
import json
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

sns.set_theme(style='whitegrid')

paper_dir = Path('./perla_extractions_for_si_analysis_sonnet5/just_paper_text')
si_dir = Path('.//perla_extractions_for_si_analysis_sonnet5/with_SI')

records = []

for paper_json in sorted(paper_dir.glob('*.json')):
    doi = paper_json.stem

    si_json = si_dir / f'{doi}-SI.json'

    if not si_json.exists():
        print(f'Missing SI for {doi}')
        continue

    with open(paper_json) as f:
        paper = json.load(f)

    with open(si_json) as f:
        si = json.load(f)

    n_paper = len(paper.get('cells', []))
    n_si = len(si.get('cells', []))

    records.append(
        {
            'doi': doi,
            'paper_only': n_paper,
            'paper_plus_si': n_si,
            'increase': n_si - n_paper,
        }
    )
# with open("10.1002--adma.202305822-SI-rerun.json") as f:
#     records.append(
#         {
#             "doi": "10.1002--adma.202305822-SI-rerun",
#             "paper_plus_si": len(json.load(f).get("cells", []))
#         }
#     )

df = pd.DataFrame(records)

print(df)

print('\nSummary')
print('----------------')
print(f'Papers: {len(df)}')
print(f'Total devices (paper): {df.paper_only.sum()}')
print(f'Total devices (+SI):   {df.paper_plus_si.sum()}')
print(f'Mean paper:            {df.paper_only.mean():.2f}')
print(f'Mean +SI:              {df.paper_plus_si.mean():.2f}')
print(f'Mean increase:         {df.increase.mean():.2f}')
print(f'Median increase:       {df.increase.median():.2f}')
print(f'Papers improved:       {(df.increase>0).sum()} / {len(df)}')


# -------------------------------------------------
# Scatter plot
# -------------------------------------------------

fig, ax = plt.subplots(figsize=(6, 6))

ax.scatter(
    df.paper_only,
    df.paper_plus_si,
    s=60,
    alpha=0.8,
)

m = max(df.paper_only.max(), df.paper_plus_si.max())
ax.plot([0, m], [0, m], 'k--', lw=1)

ax.set_xlabel('Devices extracted from paper')
ax.set_ylabel('Devices extracted from paper + SI')
ax.set_title('Effect of Supplementary Information')

plt.tight_layout()
plt.savefig('paper_vs_si_scatter.png', dpi=300)


# -------------------------------------------------
# Histogram of improvements
# -------------------------------------------------

fig, ax = plt.subplots(figsize=(7, 4))

sns.histplot(df['increase'], bins=20, ax=ax)

ax.axvline(0, color='black', linestyle='--')
ax.set_xlabel('Additional extracted devices from SI')
ax.set_ylabel('Number of papers')

plt.tight_layout()
plt.savefig('paper_vs_si_histogram.png', dpi=300)


# -------------------------------------------------
# Sorted improvement bar plot
# -------------------------------------------------

df_sorted = df.sort_values('increase')

fig, ax = plt.subplots(figsize=(10, 5))

ax.bar(range(len(df_sorted)), df_sorted['increase'])

ax.set_xlabel('Papers (sorted)')
ax.set_ylabel('Extra devices from SI')

plt.tight_layout()
plt.savefig('paper_vs_si_sorted_improvement.png', dpi=300)


# -------------------------------------------------
# Paper vs SI connected plot
# -------------------------------------------------

fig, ax = plt.subplots(figsize=(8, max(5, len(df) * 0.15)))

df2 = df.sort_values('increase')

for i, row in enumerate(df2.itertuples()):
    ax.plot(
        [0, 1],
        [row.paper_only, row.paper_plus_si],
        color='gray',
        alpha=0.5,
    )
    ax.scatter([0], [row.paper_only], color='tab:blue')
    ax.scatter([1], [row.paper_plus_si], color='tab:orange')

ax.set_xticks([0, 1])
ax.set_xticklabels(['Paper', 'Paper + SI'])
ax.set_ylabel('Extracted devices')

plt.tight_layout()
plt.savefig('paper_vs_si_slopegraph.png', dpi=300)

plt.show()
```

```python
import pandas as pd
import plotly.express as px

# Convert records to DataFrames
paper_tokens = pd.DataFrame(paper_records)
paper_tokens['dataset'] = 'Paper only'

si_tokens = pd.DataFrame(paper_si_records)
si_tokens['dataset'] = 'Paper + SI'

# Keep only DOIs present in both datasets
common_dois = set(paper_tokens['doi']).intersection(set(si_tokens['doi']))

paper_tokens = paper_tokens[paper_tokens['doi'].isin(common_dois)]

si_tokens = si_tokens[si_tokens['doi'].isin(common_dois)]

print(f'Total papers compared: {len(common_dois)}')

# Combine
tokens_df = pd.concat([paper_tokens, si_tokens], ignore_index=True)

# Reshape for plotting
plot_df = tokens_df.melt(
    id_vars=['doi', 'dataset'],
    value_vars=['prompt_tokens', 'completion_tokens'],
    var_name='token_type',
    value_name='tokens',
)

plot_df['token_type'] = plot_df['token_type'].replace(
    {
        'prompt_tokens': 'Prompt tokens',
        'completion_tokens': 'Completion tokens',
    }
)

# Plot
fig = px.bar(
    plot_df,
    x='tokens',
    y='doi',
    color='token_type',
    facet_col='dataset',
    orientation='h',
    barmode='group',
    title=f'Prompt vs completion tokens per DOI ({len(common_dois)} papers)',
)

fig.update_layout(
    height=max(600, len(common_dois) * 35),
    template='simple_white',
    xaxis_title='Tokens',
    yaxis_title='DOI',
)

fig.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Convert to DataFrames
df_main = pd.DataFrame(paper_records)
df_si = pd.DataFrame(paper_si_records)

# Keep only needed columns and rename
main_chars = df_main[['doi', 'characters']].rename(
    columns={'characters': 'paper_characters'}
)

si_chars = df_si[['doi', 'characters']].rename(columns={'characters': 'si_characters'})

# Merge by DOI
paired = main_chars.merge(si_chars, on='doi', how='inner')

# Total characters
paired['paper_plus_si'] = paired['paper_characters'] + paired['si_characters']

print(paired.head())
print('Matched papers:', len(paired))
plt.figure(figsize=(8, 8))

for _, row in paired.iterrows():
    plt.plot(
        [0, 1], [row['paper_characters'], row['paper_plus_si']], color='gray', alpha=0.4
    )

plt.scatter([0] * len(paired), paired['paper_characters'], label='Paper', s=60)

plt.scatter([1] * len(paired), paired['paper_plus_si'], label='Paper + SI', s=60)

plt.xticks([0, 1], ['Paper', 'Paper + SI'])
plt.ylabel('Characters')
plt.title('Effect of Supporting Information on pdf text ingested in pipeline length')
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```
